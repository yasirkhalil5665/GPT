# Exploration

Uses the `src/` package for training, evaluation, and generation. This notebook is for experimentation and visualization only — no model/data logic lives here.

## Setup

```
import sys
sys.path.append("..")

import matplotlib.pyplot as plt

from src.config import config
from src.train import train
from src.generate import generate
```

## Train

```
train_losses, test_losses = train()
```

## Loss Curve

```
plt.figure(figsize=(8, 5))
plt.plot(train_losses, label="train_loss")
plt.plot(test_losses, label="test_loss")
plt.xlabel(f"Evaluation step (every {config.eval_interval} iterations)")
plt.ylabel("Loss")
plt.title("Training Progress")
plt.legend()
plt.show()
```
